```python
costing = 0
dis = 0
run=0
print("="*50)
print("\t SMARTCAMPUS UTILITY & ACCESS PASS GENERATOR")
print("="*50)
asad = int(input("enter Category(1: Student,2: teacher): "))
Pass = 0
disc = 0
if asad==1:
    Asad = input("Enter SubCategory(UG/PG): ")
    if Asad.lower() == 'ug':
        costing += 500
        Pass = 500
    elif Asad.lower() == 'pg':
        costing += 350
        Pass = 350
    else:
        run=1
    if run==0:
        cgpa = float(input("Enter Student CGPA: "))
        if cgpa>=8.5:
            costing = costing - (costing*0.2)
            disc = Pass*0.2
            dis = 20
        elif cgpa<8.50 and cgpa>7.5:
            costing = costing - (costing*0.1)
            disc = Pass*0.1
            dis = 10
elif asad==2:
    Asad = int(input("Enter SubCatagory(1: Resident teacher,2: Visiting teacher): "))
    if Asad==1:
        costing += 800
        Pass = 800
    elif Asad == 2:
        costing += 1200
        Pass = 120
    else:
        run=1
    if run==0:
        service = int(input("Enter Year of work experience: "))
        if service>10:
            costing = costing - (costing*0.15)
            disc=Pass*0.15
            dis = 15
else:
    run=1
if run==0:
    parking_fee = 0
    parking = int(input("Select Parking Permit (0: None,2: Motor bikes,4: car): "))
    if parking==2:
        costing += 200
        parking_fee = 200
    elif parking==4:
        costing += 600
        parking_fee = 600
    else:
        run=1
if run==0:
    elebill = 0
    Total = costing
    fix=0
    unit = int(input("Enter  your monthly electricity consumption (in KWH): "))
    if unit>500:
        elebill = (unit - 500)*10 
        fix=250
        elebill += (300+1000+1500+fix+300)
        costing += elebill
    elif unit>300:
        elebill = (unit - 300)*7.50
        fix=150
        elebill += (1000+300+fix+150)
        costing += elebill
    elif unit>100:
        elebill = (unit - 100)*5
        fix = 100
        elebill += (300+fix+50)
        costing += elebill
    else:
        elebill = unit*3
        fix = 50
        elebill += fix
        costing += elebill
if run==0:
    print('-'*50)
print("CALCULATED INVOICE BREAKDOWN")
print('-'*50)
print(f"Base Access Pass Fee   :₹{Pass}")
print(f"Merit Discount ({dis}%)   :-₹{disc}")
print(f"Parking Fee ({parking}-Wheeler)   :₹{parking_fee}")
if asad==1 and parking==4:
    Total+=150
    costing += 150
    print("Student Peak Surcharge   :150")
    print(f"Net Pass & Parking Total   :₹{Total}")
    print("-"*50)
    print(f"Electricity Bill({unit} kWh) :₹{elebill} ",end="")
    print(f"(Slab calculate + Fixed Charge {fix})")
    print("-"*50)
    print(f"TOTAL MONTHLY PAYABLE   :₹{costing}")
    print("="*50)
```
